```julia
#1
A = Set([0, 3, 4, 9])
B = Set([1, 3, 4, 7])
C = Set([0, 1, 2, 4, 7, 8, 9])

P = union(
    intersect(A, B),
    intersect(A, C),
    intersect(B, C)
)

println("P = ", P)
```

```julia
#2
A = Set([1, 2, 3, 4])
B = Set([3, 4, 5, 6])

println("Объединение: ", union(A, B))
println("Пересечение: ", intersect(A, B))
println("Разность A \\ B: ", setdiff(A, B))
println("3 ∈ A: ", 3 in A)

# строки
S1 = Set(["Julia", "Python", "C++"])
S2 = Set(["Julia", "Java", "C++"])

println("Объединение строк: ", union(S1, S2))
println("Пересечение строк: ", intersect(S1, S2))

# символы
C1 = Set(['a', 'b', 'c'])
C2 = Set(['b', 'c', 'd'])

println("Объединение символов: ", union(C1, C2))
println("Пересечение символов: ", intersect(C1, C2))
```
