```julia
#8
function outer(x,y,operation)
    [operation(a,b) for a in x, b in y]
end
outer([1,2], [10,20,30], +)
```

```julia
A1 = outer(0:4, 0:4, +)
display(A1)

A2 = outer(0:4, 0:4, ^)
display(A2)

A3 = [mod((i-1)+(j-1), 5) for i in 1:5, j in 1:5]
display(A3)

A4 = [mod(i+j, 10) for i in 0:9, j in 0:9]
display(A4)

A5 = [mod(i-j, 9) for i in 0:8, j in 0:8]
display(A5)
```

```julia
#10
using Random
Random.seed!(1234) 
M = rand(1:10, 6, 10)
display(M)

N = 4
counts_greater_N = [count(>(N), M[i, :]) for i in 1:size(M, 1)]
println("Количество элементов > $N в каждой строке:")
display(counts_greater_N)

target = 7
rows_with_two_targets = [i for i in 1:size(M, 1) if count(==(target), M[i, :]) == 2]
println("Строки, где число $target встречается ровно дважды: ", rows_with_two_targets)

K = 75
column_pairs = Tuple{Int, Int}[]
for i in 1:size(M, 2)-1, j in i+1:size(M, 2)
    if sum(M[:, i]) + sum(M[:, j]) > K
        push!(column_pairs, (i, j))
    end
end
println("Пары столбцов, сумма всех элементов которых > $K:")
display(column_pairs)
```
