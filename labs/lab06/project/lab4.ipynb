```julia
# Задаём квадратную матрицу 3x3 со случайными значениями:
A = rand(3, 3)
# Задаём единичный вектор:
x = fill(1.0, 3)
# Задаём вектор b:
b = A*x
# Решение исходного уравнения получаем с помощью функции \
# (убеждаемся, что x - единичный вектор):
A\b
```

```julia
# LU-факторизация:
Alu = lu(A)
```

```julia
# Матрица перестановок:
Alu.P
# Вектор перестановок:
Alu.p
# Матрица L:
Alu.L
# Матрица U:
Alu.U
# Решение СЛАУ через матрицу A:
A\b
# Решение СЛАУ через объект факторизации:
Alu\b
# Детерминант матрицы A:
det(A)
# Детерминант матрицы A через объект факторизации:
det(Alu)
```

```julia
# QR-факторизация:
Aqr = qr(A)
```

```julia
# Матрица Q:
Aqr.Q
# Матрица R:
Aqr.R
# Проверка, что матрица Q - ортогональная:
Aqr.Q'*Aqr.Q
# Симметризация матрицы A:
Asym = A + A'
# Спектральное разложение симметризованной матрицы:
AsymEig = eigen(Asym)
# Собственные значения:
AsymEig.values
#Собственные векторы:
AsymEig.vectors
# Проверяем, что получится единичная матрица:
inv(AsymEig)*Asym
```

```julia
# 4.4.2

using LinearAlgebra

function solve_system(A, b)
    rA = rank(Float64.(A))
    rAb = rank(Float64.(hcat(A, b)))

    if rA < rAb
        return "Решений нет"
    elseif rA < size(A, 2)
        return "Бесконечно много решений"
    else
        return A \ b
    end
end
```

```julia
# a)
A_a = [1 1; 1 -1]
b_a = [2, 3]
println("a) ", solve_system(A_a, b_a))
```

```julia
#4.4.3

using LinearAlgebra

A1 = [1.0 -2; -2 1]
A2 = [1.0 -2; -2 3]
A3 = [1.0 -2 0; -2 1 2; 0 2 0]

for (name, A) in [("a", A1), ("b", A2), ("c", A3)]
    F = eigen(Symmetric(A))
    Q = F.vectors
    D = Diagonal(F.values)

    println("\nМатрица ", name)
    println("Собственные значения:")
    display(F.values)
    println("Диагональная матрица D:")
    display(D)
    println("Матрица собственных векторов Q:")
    display(Q)
    println("Проверка A = Q*D*Q': ",
            isapprox(A, Q * D * Q'))
end
```

```julia
#3

using LinearAlgebra

A = [
    140  97  74 168 131
     97 106  89 131  36
     74  89 152 144  71
    168 131 144  54 142
    131  36  71 142  36
]

# Собственные значения симметричной матрицы
values = eigvals(Symmetric(float.(A)))

# Диагональная матрица собственных значений
D = Diagonal(values)

# Нижнетреугольная матрица из A
L = tril(A)

println("Собственные значения:")
display(values)

println("Диагональная матрица D:")
display(D)

println("Нижнетреугольная матрица L:")
display(L)

println("\nВремя вычисления собственных значений:")
@time eigvals(Symmetric(float.(A)))

println("\nВремя создания диагональной матрицы:")
@time Diagonal(values)

println("\nВремя создания нижнетреугольной матрицы:")
@time tril(A)
```

```julia
#4.1

using LinearAlgebra

A1 = [1.0 2; 3 4]
A2 = (1 / 2) * A1
A3 = (1 / 10) * A1

function check_productivity(A)
    M = I - A

    if iszero(det(Matrix(M)))
        println("Матрица I-A вырождена.")
        return false
    end

    C = inv(Matrix(M))
    println("(I-A)^(-1) = ")
    display(C)

    productive = all(C .>= -1e-10)
    println("Продуктивная: ", productive)
    return productive
end

println("a)")
check_productivity(A1)

println("b)")
check_productivity(A2)

println("c)")
check_productivity(A3)
```

```julia
#4.2

using LinearAlgebra

B1 = [1.0 2; 3 1]
B2 = (1 / 2) * B1
B3 = (1 / 10) * B1

function inverse_criterion(A)
    M = Matrix(I, size(A, 1), size(A, 2)) - A

    if abs(det(M)) < 1e-10
        println("I-A вырождена: обратной матрицы нет.")
        return false
    end

    C = inv(M)
    println("(I-A)^(-1) = ")
    display(C)

    productive = all(C .>= -1e-10)
    println("Продуктивная: ", productive)
    return productive
end

println("a)")
inverse_criterion(B1)

println("b)")
inverse_criterion(B2)

println("c)")
inverse_criterion(B3)
```

```julia
#4.3

using LinearAlgebra

B1 = [1.0 2; 3 1]
B2 = (1 / 2) * B1
B3 = (1 / 10) * B1

B4 = [
    0.1 0.2 0.3
    0.0 0.1 0.2
    0.0 0.1 0.3
]

function spectral_criterion(A)
    values = eigvals(A)
    rho = maximum(abs.(values))

    println("Собственные значения: ", values)
    println("Спектральный радиус: ", rho)
    println("Продуктивная: ", rho < 1)

    return rho < 1
end

println("a)")
spectral_criterion(B1)

println("b)")
spectral_criterion(B2)

println("c)")
spectral_criterion(B3)

println("d)")
spectral_criterion(B4)
```
