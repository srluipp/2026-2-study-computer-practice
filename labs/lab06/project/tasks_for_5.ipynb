```julia
#1
using Plots
x = range(0, 2π, length=100)
y = sin.(x)

p1 = plot(x, y, title="Линия", label="sin(x)")

p2 = scatter(x, y, title="Точечный", label="sin(x)")

p3 = bar(x[1:10:end], y[1:10:end],
         title="Столбчатый", label="sin(x)")

p4 = histogram(y, title="Гистограмма", label="sin(x)")

# График с заполненной областью
p5 = plot(x, y, fillrange=0, fillalpha=0.3,
          title="Область", label="sin(x)")

# Ступенчатый график
p6 = plot(x, y, seriestype=:steppre,
          title="Ступенчатый", label="sin(x)")

plot(p1, p2, p3, p4, p5, p6,
     layout=(2,3), size=(1000,700))
```

```julia
#2
x = range(0, 2π, length=200)
y = sin.(x)

styles = [:solid, :dash, :dot, :dashdot, :dashdotdot]

p = plot(title="Разные стили линий",
         xlabel="x", ylabel="sin(x)")

for style in styles
    plot!(p, x, y,
          linestyle=style,
          label=string(style))
end

plot!(p, x, y, linestyle=:solid,
      marker=:circle, label="Круги")

plot!(p, x, y, linestyle=:solid,
      marker=:square, label="Квадраты")

plot!(p, x, y, linestyle=:solid,
      marker=:diamond, label="Ромбы")

display(p)
```

```julia
#3
f(x) = π * x^2 / log(x)

x1 = range(0.2, 0.99, length=300)
x2 = range(1.01, 4.0, length=300)

p = plot(x1, f.(x1),
    color=:red, label="y = πx² / ln(x)",
    xlabel="Аргумент x",
    ylabel="Значение y",
    title="График функции",
    foreground_color_subplot=:green,
    foreground_color_legend=:green,
    guidefont=font(12, "Arial"),
    tickfont=font(10, "Arial"),
    titlefont=font(14, "Arial"),
    margin=8Plots.mm,
    xticks=0:0.5:4,
    yticks=-20:20:100)

plot!(p, x2, f.(x2), color=:red, label="")
display(p)
```

```julia
#5
x = 3:0.1:6

y1 = π .* x
y2 = exp.(x) .* cos.(x)

p = plot(x, y1,
    color=:blue,
    label="πx",
    xlabel="x",
    ylabel="y",
    title="Две функции",
    legend=:topright,
    grid=true)

plot!(p, x, y2,
    color=:red,
    label="exp(x) cos(x)")

display(p)
```

```julia
#10
function hypocycloid(k, t)
    R = k
    r = 1.0

    x = (R-r).*cos.(t) .+
        r.*cos.(((R-r)/r).*t)

    y = (R-r).*sin.(t) .-
        r.*sin.(((R-r)/r).*t)

    return x, y
end

ks = [3//1, 4//1, 5//2, 7//3]

for k in ks
    kf = Float64(k)
    period = 2π * denominator(k)

    anim = @animate for tmax in range(0.1, period, length=100)
        t = range(0, tmax, length=500)
        x, y = hypocycloid(kf, t)

        plot(x, y,
            aspect_ratio=:equal,
            title="Гипоциклоида k=$k",
            xlabel="x",
            ylabel="y",
            label="k=$k")
    end

    folder = "hypocycloid_k_$(k)"
    mkpath(folder)

    gif(anim, joinpath(folder, "animation.gif"), fps=15)
end
```

```julia
#11

function epicycloid(k, t)
    R = k
    r = 1.0

    x = (R+r).*cos.(t) .-
        r.*cos.(((R+r)/r).*t)

    y = (R+r).*sin.(t) .-
        r.*sin.(((R+r)/r).*t)

    return x, y
end

ks = [3//1, 4//1, 5//2, 7//3]

for k in ks
    kf = Float64(k)
    period = 2π * denominator(k)

    anim = @animate for tmax in range(0.1, period, length=100)
        t = range(0, tmax, length=500)
        x, y = epicycloid(kf, t)

        plot(x, y,
            aspect_ratio=:equal,
            title="Эпициклоида k=$k",
            xlabel="x",
            ylabel="y",
            label="k=$k")
    end
    folder = "epicycloid_k_$(k)"
    mkpath(folder)

    # Сохраняем анимацию
    gif(anim, joinpath(folder, "animation.gif"), fps=15)
end
```
